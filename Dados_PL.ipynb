```python
#Utilizando as bibliotecas Pandas, Seaborn e Matplotlib:
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Dados gerados como dicionário:
data = {
    'Time': ['Manchester City', 'Liverpool', 'Arsenal', 'Chelsea', 'Tottenham', 'Manchester United'],
    'Gols Marcados': [80, 70, 75, 68, 65, 60],
    'Gols Sofridos': [30, 35, 40, 45, 38, 42],
    'Vitórias': [25, 22, 23, 20, 18, 19],
    'Empates': [5, 8, 7, 10, 6, 9],
    'Derrotas': [3, 4, 5, 6, 7, 6],
    'Pontos': [80, 74, 76, 70, 69, 66]
}
```

```python
dados_pl = pd.DataFrame(data)
```

```python
dados_pl
```

## Cálculo Saldo de Gols:

```python
#Saldo de gols é composto pela diferença entre gols marcados e gols sofridos:
dados_pl['Saldo Gols'] = dados_pl['Gols Marcados'] - dados_pl['Gols Sofridos']
```

```python
dados_pl
```

## Ordenando times por pontos:

```python
dados_pl.sort_values(by = 'Pontos', axis = 0, ascending = False, inplace = True) #Ordenando por pontos através do parâmetro 'by'
```

```python
dados_pl
```

```python
#Função para retornar time com maior pontuação:
for i in range(len(dados_pl['Time'])):
    if dados_pl['Pontos'][i] == dados_pl['Pontos'].max():
        nome_time = dados_pl['Time'][i]

nome_time
```

```python
print(f'O time com maior pontuação na Premier League é: {nome_time}')
```

## Taxa de Vitórias (N° vitórias / Total Jogos):

```python
#Criando coluna total de jogos:
dados_pl['Total de Jogos'] = dados_pl['Vitórias'] + dados_pl['Empates'] + dados_pl['Derrotas']
```

```python
dados_pl
#O total de jogos pode varia de time para time devido competições internacionais
```

```python
#Criando coluna 'Taxa de Vitórias':
dados_pl['Taxa de Vitórias (%)'] = round(((dados_pl['Vitórias'] / dados_pl['Total de Jogos']) * 100), 2) #Arredondando porcentagem
```

```python
dados_pl
```

```python
#Função para calcular time com maior taxa de vitórias:
for i in range(len(dados_pl['Time'])):
    if dados_pl['Taxa de Vitórias (%)'][i] == dados_pl['Taxa de Vitórias (%)'].max():
        maior_taxa = dados_pl['Time'][i]

maior_taxa
```

```python
print(f'O clube com maior taxa de vitórias é: {maior_taxa}')
```

## Gráfico gols marcados:

```python
#Gráficos de barras (matplotlib):
x = dados_pl['Time']
y = dados_pl['Gols Marcados']

ax = plt.figure(figsize = (10, 6))

ax = plt.bar(x, y)
ax = plt.xlabel('Times')
ax = plt.ylabel('Gols Marcados')
ax = plt.title('Melhores Ataques')
ax = plt.legend()
ax = plt.show()
```

## Relação Gols Marcados vs. Pontuação

```python
dados_pl.columns
```

```python
#Correlação: variável_df_x.corr(variável_df_y), onde ambas as variáveis estão sob análise de correlação
correlacao = dados_pl['Gols Marcados'].corr(dados_pl['Pontos'])
```

```python
print(f'A relação entre gols marcados e pontos é: {round(correlacao, 2)}')
print('Ou seja: existe uma relação direta entre pontos e gols marcados.')
```

## Aproveitamento do Manchester United com Bruno Fernandes em campo:

```python
import pandas as pd
```

```python
dados = {
    'jogo_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'data': ['2024-08-01', '2024-08-05', '2024-08-12', '2024-08-20', '2024-08-27', '2024-09-02', '2024-09-09', '2024-09-15'],
    'bruno_fernandes_em_campo': [True, False, True, True, False, True, False, True],
    'vitoria': [True, False, True, False, False, True, True, True],
    'empate': [False, True, False, False, True, False, False, False]
}
```

```python
aproveitamento_bf = pd.DataFrame(dados)
```

```python
aproveitamento_bf
```

```python
aproveitamento_bf.info() #Conclui-se que data não está no formato correto
```

```python
from datetime import datetime
```

```python
#Necessária a conversão do tipo string para datetime na coluna 'data'
aproveitamento_bf['data'] = pd.to_datetime(aproveitamento_bf['data'])
```

```python
aproveitamento_bf.info()
```

```python
#Determine o aproveitamento de pontos do Manchester United em partidas onde Bruno Fernandes está em campo:
aproveitamento_bf
```

```python
quantidade_jogos = len(aproveitamento_bf['jogo_id'])
quantidade_jogos #Número de partidas
```

```python
partidas_bf = 0
soma_pontos = 0
for i in range(len(aproveitamento_bf['bruno_fernandes_em_campo'])):
    if aproveitamento_bf['bruno_fernandes_em_campo'][i] == True:
        partidas_bf = partidas_bf + 1
        if ((aproveitamento_bf['vitoria'][i] == True) & (aproveitamento_bf['empate'][i] == False)): #Vitória
            soma_pontos = soma_pontos + 3
        elif ((aproveitamento_bf['vitoria'][i] == False) & (aproveitamento_bf['empate'][i] == True)): #Empate
             soma_pontos = soma_pontos + 1
        elif ((aproveitamento_bf['vitoria'][i] == False) & (aproveitamento_bf['empate'][i] == False)): #Derrota
            soma_pontos = soma_pontos + 0
        else: #Qualquer outro caso (inexistente)
            print('Erro no Conjunto de Dados') 
            break
    else:
        partidas_bf = partidas_bf + 0
```

```python
partidas_bf
```

```python
soma_pontos
```

```python
#Sabendo-se quem cada partida, o máximo de pontos possível é 3:
total_pontos = partidas_bf * 3

#Dessa forma, checa-se o aproveitamento do clube enquanto Bruno Fernando esteve em campo:
print(f'O aproveitamento de pontos com Bruno Fernandes em campo é de {round((soma_pontos / total_pontos) * 100, 2)}%')
```

## Média de gols tomados por partida quando o defensor Raphael Varane não joga:

```python
dados = {
    'jogo_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'data': ['2024-08-01', '2024-08-05', '2024-08-12', '2024-08-20', '2024-08-27', '2024-09-02', '2024-09-09', '2024-09-15'],
    'raphael_varane_em_campo': [True, False, True, False, True, False, True, False],
    'gols_sofridos': [1, 3, 0, 2, 1, 4, 1, 3]
}
```

```python
dados_varane = pd.DataFrame(dados)
```

```python
dados_varane
```

```python
dados_varane.info()
```

```python
#Convertendo data para datetime:
dados_varane['data'] = pd.to_datetime(dados_varane['data'])
```

```python
dados_varane.info()
```

```python
#Calcule a média de gols sofridos por partida quando Raphael Varane não está em campo:
dados_varane
```

```python
jogos_sem_varane = 0
gols_tomados = 0

for i in range(len(dados_varane['raphael_varane_em_campo'])):
    if dados_varane['raphael_varane_em_campo'][i] == False:
        jogos_sem_varane = jogos_sem_varane + 1
        gols_tomados = gols_tomados + dados_varane['gols_sofridos'][i]

print(f'Varane não jogou {jogos_sem_varane} partidas. Em suas ausências, a equipe levou {gols_tomados} gols. ')
print(f'Com isso, a média de gols tomados sem o francês em campo foi de {round((gols_tomados / jogos_sem_varane), 2)} por partida.')
```

## Relação entre posse de bola e vitórias com Marcus Rashford no time:

```python
dados = {
    'jogo_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'data': ['2024-08-01', '2024-08-05', '2024-08-12', '2024-08-20', '2024-08-27', '2024-09-02', '2024-09-09', '2024-09-15'],
    'marcus_rashford_em_campo': [True, True, False, True, False, True, True, False],
    'posse_de_bola': [55, 48, 52, 63, 45, 57, 60, 49],
    'vitoria': [True, False, False, True, False, True, True, False]
}
```

```python
dados_rashford = pd.DataFrame(dados)
```

```python
dados_rashford
```

```python
dados_rashford.info()
```

```python
dados_rashford['data'] = pd.to_datetime(dados_rashford['data'])
```

```python
dados_rashford.info() #Dqdos convertidos
```

```python
#Verifique se há alguma correlação entre a posse de bola e as vitórias quando Marcus Rashford está jogando:
dados_rashford
```

```python
posse_com_marcus = []
resultado_com_marcus = []

for i in range(len(dados_rashford['marcus_rashford_em_campo'])):
    if (dados_rashford['marcus_rashford_em_campo'][i] == True):
        posse_com_marcus.append(dados_rashford['posse_de_bola'][i])
        resultado_com_marcus.append(dados_rashford['vitoria'][i])
```

```python
dicionario_rashford = {'Posse_Marcus': posse_com_marcus, 'Resultado_Marcus': resultado_com_marcus}
```

```python
correlacao_rashford = pd.DataFrame(dicionario_rashford)
correlacao_rashford
```

```python
correlacao = correlacao_rashford['Posse_Marcus'].corr(correlacao_rashford['Resultado_Marcus'])
```

```python
print(f'Existe uma correlação de {round(correlacao, 2)} entre posse posse de vitórias quando Marcus Rashford atua.')
```

## Média de assistências por jogo quando o time tem menos de 50% de posse de bola:

```python
dados = {
    'jogo_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'data': ['2024-08-01', '2024-08-05', '2024-08-12', '2024-08-20', '2024-08-27', '2024-09-02', '2024-09-09', '2024-09-15'],
    'posse_de_bola': [55, 48, 49, 63, 45, 57, 40, 49],
    'assistencias': [2, 1, 1, 3, 0, 2, 0, 1]
}
```

```python
dados_assistencia = pd.DataFrame(dados)
```

```python
dados_assistencia
```

```python
dados_assistencia.info()
```

```python
#Corrigindo parâmetro de datas:
dados_assistencia['data'] = pd.to_datetime(dados_assistencia['data'])
```

```python
dados_assistencia.info()
```

```python
#Calcule a média de assistências por jogo em partidas onde o Manchester United teve menos de 50% de posse de bola:
dados_assistencia
```

```python
jogos_posse_menor_cinquenta = 0
quantidade_assistencias = 0

for i in range(len(dados_assistencia['posse_de_bola'])):
    if (dados_assistencia['posse_de_bola'][i] < 50):
        jogos_posse_menor_cinquenta = jogos_posse_menor_cinquenta + 1
        quantidade_assistencias = quantidade_assistencias + dados_assistencia['assistencias'][i]
```

```python
assistencias_por_partida = quantidade_assistencias / jogos_posse_menor_cinquenta
assistencias_por_partida
```

```python
print(f'A média de assistências por partida quando o time tem menos que 50% de posse de bola é de {assistencias_por_partida}.')
```

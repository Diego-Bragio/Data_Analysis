```python
import pandas as pd
```

```python
dados = {
    'temporada': [2010, 2011, 2012, 2013, 2014, 2015, 2016],  # Exemplo de anos de participação
    'equipe': ['Nottingham Forest', 'Nottingham Forest', 'Nottingham Forest', 
               'Aston Villa', 'Aston Villa', 'Aston Villa', 'Aston Villa'],
    'posse_bola': [58.2, 55.1, 57.3, 60.5, 62.1, 59.4, 56.8],  # Em %
    'gols_marcados': [10, 12, 14, 16, 18, 13, 15],  # Total de gols marcados
    'gols_sofridos': [8, 9, 10, 11, 10, 9, 8],  # Total de gols sofridos
    'vitorias': [4, 5, 6, 7, 5, 6, 5],  # Número de vitórias
    'empates': [2, 1, 2, 1, 3, 2, 3],  # Número de empates
    'derrotas': [0, 1, 1, 1, 1, 1, 2],  # Número de derrotas
    'classificacao': [2, 1, 2, 4, 2, 3, 2],  # Classificação final
    'campeao': [False, False, False, False, False, False, True],  # Campeão ou não
}
```

```python
villa_forest = pd.DataFrame(dados)
```

```python
villa_forest
```

## Calcule a média de gols marcados, gols sofridos e posse de bola para cada time ao longo das 7 temporadas analisadas:

```python
forest = villa_forest.query('equipe == "Nottingham Forest"')
forest
```

```python
media_posse = forest['posse_bola'].mean()
media_marcado = forest['gols_marcados'].mean()
media_sofrido = forest['gols_sofridos'].mean()

print(f'Nottingham Forest teve uma média de {round(media_posse, 2)}% de posse de bola, {media_marcado} de gols marcados e \
{media_sofrido} de gols sofridos.')
```

```python
villa = villa_forest.query('equipe == "Aston Villa"')
villa
```

```python
media_posse_villa = villa['posse_bola'].mean()
media_marcado_villa = villa['gols_marcados'].mean()
media_sofrido_villa = villa['gols_sofridos'].mean()

print(f'Aston Villa teve uma média de {round(media_posse_villa, 2)}% de posse de bola, {media_marcado_villa} de gols marcados e \
{media_sofrido_villa} de gols sofridos.')
```

## Qual time teve maior consistência em termos de gols marcados e gols sofridos? Identifique a temporada mais equilibrada (em termos de gols marcados e sofridos) para cada time:

```python
villa_forest['saldo_gols'] = villa_forest['gols_marcados'] - villa_forest['gols_sofridos']
villa_forest
```

```python
#Verificar Maior consistência, é verificar o saldo de gols. Assim:
novo_forest = villa_forest.query('equipe == "Nottingham Forest"')
novo_forest
```

```python
novo_forest['saldo_gols'].sum() #Saldo de gols Nottingham Forest
```

```python
novo_villa = villa_forest.query('equipe == "Aston Villa"')
novo_villa.reset_index(drop = True, inplace = True)
```

```python
novo_villa['saldo_gols'].sum() #Saldo de gols Aston Villa
```

```python
#Um saldo de gols equilibrado seria próximo a zero. Assim:
menor_saldo_forest = novo_forest['saldo_gols'].min()
for i in range(len(novo_forest)):
    if novo_forest['saldo_gols'][i] == menor_saldo_forest:
        temporada_equilibrada_forest = novo_forest['temporada'][i]


menor_saldo_villa = novo_villa['saldo_gols'].min()
for i in range(len(novo_villa)):
    if novo_villa['saldo_gols'][i] == menor_saldo_villa:
        temporada_equilibrada_villa = novo_villa['temporada'][i]

print(f'A temporada mais equilibrada do Nottingham Forest foi em {temporada_equilibrada_forest} com {menor_saldo_forest} gols de saldo.')
print(f'A temporada mais equilibrada do Aston Villa foi em {temporada_equilibrada_villa} com {menor_saldo_villa} gols de saldo.')  
```
